#### Reading CSV Files with dtype 
###### The project contains four raw CSV files: routes, stops, validations and vehicles. The data will be loaded with explicit data types so that IDs remain strings.

```python
stops = pd.read_csv("data-analysis-bootcamp-capstone-project/data/raw/stops.csv",
                    dtype={"stop_id":"string",
                           "stop_name":"string",
                           "district":"string",
                           "zone":"string",
                           "has_shelter":"string"
                           }
                           )
stops.dtypes
```

```python
validations = pd.read_csv("data-analysis-bootcamp-capstone-project/data/raw/validations.csv",
                          dtype={"validation_id":"string",
                                 "validation_ts":"string",
                                 "route_id":"string",
                                 "vehicle_id":"string",
                                 "stop_id":"string",
                                 "card_id":"string",
                                 "fare_type":"string",
                                 "fare_amount":"float64",
                                 "passenger_count": "float64",
                                 "transfer_flag": "float64",
                                "transfer_flag": "string"
                                }
)
validations.dtypes
```

```python
vehicles = pd.read_csv("data-analysis-bootcamp-capstone-project/data/raw/vehicles.csv",
                       dtype={"vehicle_id":"string",
                              "vehicle_type":"string",
                              "capacity": "float64",
                              "year_manufactured":"int64",
                              "depot":"string"}
                              )
vehicles.dtypes
```

### Routes
#### These routes profiles below are used to show us how many rows and columns each table has, data types and missing values and also the memory usage.

```python
print("Rows", routes.shape[0])

print("Columns", routes.shape[1])

print("Data type:")
print(routes.dtypes)

print("Null Counts:")
print(routes.isna().sum())

print("Null percentage:")
print(routes.isna().mean()* 100)

print("Memory usage:")
print(routes.memory_usage(deep= True).sum(),"bytes")

print("First rows:")
display(routes.head())
```

### Step 4 looking with value.counts to see if something looks wrong

##### In the routes[mode] column we are seeing values that appear to represent the same transport mode, but are written differntly.

```python
routes["mode"].value_counts()
```

##### The stops[has_shelter] column contains mixed representations of the same concept: Y\N and 1\0 values are present.

```python
stops["has_shelter"].value_counts()
```

##### In the validations[fare_type] column we can also notice that we are having values that are the same but written differently (for example: ADULT, adult and Adult, or Senior, senior and Senior Citizen...). This will be changed during the cleaning stage.

##### In the validations[transfer_flag] column we have again mixed representations of the same concept: Y\N 1\0 and TRUE\FALSE. They are all the same.

```python
validations["fare_type"].value_counts()
```

```python
validations["transfer_flag"].value_counts()
```

#### The raw DataFrames will now be saved unchanged to the data/interim folder for the next notebook

```python
from pathlib import Path

interim_path = Path("data-analysis-bootcamp-capstone-project/data/interim")
interim_path.mkdir(parents=True, exist_ok=True)

routes.to_csv(interim_path/"routes.csv", index=False)
stops.to_csv(interim_path/"stops.scv", index=False)
validations.to_csv(interim_path/"validations.scv", index=False)
vehicles.to_csv(interim_path/"vehicles.scv", index=False)

print("All four raw datasets  were saved to data/interim/")
```
